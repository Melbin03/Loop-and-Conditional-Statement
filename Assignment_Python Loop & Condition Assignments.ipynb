```python
#Python Loop & Condition Assignments
#Even Numbers Printer Write a for loop to print all even numbers between 1 and 100.
print("Even Numbers between 1 and 100")
for i in range(1, 101):
    if i % 2 == 0:
        print(i)
```

```python
#Sum of Natural Numbers Use a while loop to calculate the sum of the first 50 natural numbers.
i=1
total=0
while i <=50:
    total=total+i
    i=i+1
print("Sum of first 50 natural numbers:",total)
    
```

```python
#Factorial Calculation Write a program using a loop to calculate the factorial of a given number n.
n=int(input("Enter the number"))
fact=1
if n < 0:
    print("Enter positive number to find factorial")
else:
    for i in range(1,n+1):
        fact=fact*i
print(f'The factorial of {n} is {fact}')
```

```python
#Reverse String Use a while loop to reverse the string "assignment" without using slicing.
s="assignment"
rev=""
i=len(s)-1
while i >= 0:
    rev=rev+s[i]
    i=i-1
print(f'Reverse of the string "assignment" is {rev}')
    
    
```

```python
#Count Vowels Write a loop with conditional statements to count the number of vowels in the string "programming language".
s="programming language"
count=0
for i in s:
    if i in 'aeiou':
        count=count+1
print(f'The number of vowels in string "programming language" is {count}') 
    
    
```

```python
#Palindrome Concept
text=input("Enter the text or number")
rev=""
for i in range(len(text)):
    rev=text[i]+rev
if rev== text:
    print("Palindrome")
else:
    print("Not a Palindrome")
```

```python
#Palindrome Concept

text = input("Enter a word or number: ")
rev = ""
for i in text:
    rev = i + rev

if text == rev:
    print(text, "is a palindrome")
else:
    print(text, "is not a palindrome")
```

```python
#Armstrong Number
num=int(input("Enter the number"))
num_1=num
n_digits=len(str(num))
total=0
while num>0:
    digit=num%10
    total=total+digit**n_digits
    num=num//10
if total==num_1:
    print(f'The number {num_1} is an Armstrong number')
else:
    print(f'The number {num_1} is not an armstrong number')
```

```python
#Print all prime numbers between 1 and 100.
print("All prime numbers between 1 and 100")
for num in range(1,101):
    count=0
    for i in range(1, num+1):
        if num % i == 0:
            count = count + 1
    if count == 2:
        print(num)
```

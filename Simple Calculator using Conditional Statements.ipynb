```python
print("Simple Calculator")
print("Choose Operation: +, -, *, /")
ops=input("Enter Operation: ")
num_1=int(input("Enter first number: "))
num_2=int(input("Enter second number: "))
if ops == '+':
    add=num_1+num_2
    print("Result:",add)
elif ops == '-':
    sub=num_1-num_2
    print("Result:",sub)
elif ops == '*':
    mult=num_1*num_2
    print("Result:",mult)
elif ops == '/':
    if num_2 == 0:
        print("Zero Division Error")
    else:
        div=num_1/num_2
        print("Result:",div)
else:
    print("Check your Operation. Select Operation from (+, -, *, /)")
```
